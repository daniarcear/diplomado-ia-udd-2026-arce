## Ejercicio 1 — Ver embeddings de tu propio proyecto

Los embeddings son la forma en que la IA "entiende" palabras: las convierte en coordenadas matemáticas donde los conceptos parecidos quedan cerca.

Vamos a comprobarlo con palabras de tu proyecto.

```python
from sentence_transformers import SentenceTransformer, util
# Modelo pequeño, se descarga la primera vez (30 seg aprox)
modelo = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
print("Modelo cargado.")
```

### TU TURNO — modifica esta lista con palabras de tu proyecto

```python
# Reemplaza estas palabras con las de tu propio proyecto
# Mezcla conceptos cercanos y lejanos para ver la geometria
palabras = [
    # cluster diseno
    "cafe", "cliente", "cafeteria", "branding",
    # cluster IA
    "modelo", "dataset", "prompt", "algoritmo",
    # cluster usuario
    "usuario", "experiencia", "interfaz", "accesibilidad",
]

# Comparar cada palabra con las demas
print("MATRIZ DE SIMILITUD (cuanto mas alto, mas parecidos):\n")
print(f"{'':15s}", end="")
for p in palabras[:12]:
    print(f"{p[:8]:>10s}", end="")
print()
for p1 in palabras[:12]:
    print(f"{p1[:14]:15s}", end="")
    for p2 in palabras[:12]:
        emb1 = modelo.encode(p1, convert_to_tensor=True)
        emb2 = modelo.encode(p2, convert_to_tensor=True)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"{sim:>10.2f}", end="")
    print()
```

**Lo que estás viendo**: cada celda muestra qué tan "parecidas" son dos palabras para el modelo. Un valor cercano a 1.00 = casi lo mismo. Cercano a 0 = distintos. Negativo = opuestos.

## Ejercicio 2 — Modelo analítico (clasificador de sentimiento)

Este modelo NO genera nada. Solo clasifica: le das un texto, te dice si es positivo o negativo.

```python
from transformers import pipeline
clasificador = pipeline("sentiment-analysis",
                       model="nlptown/bert-base-multilingual-uncased-sentiment")
print("Clasificador listo.")
```

### TU TURNO — prueba con frases de tu proyecto o inventadas

```python
frases_a_clasificar = [
    "Este cafe me parece fuerte, dulce y memorable.",
    "El cafe esta totalmente quemado, no sirve para nada.",
    "El cafe cumple con lo esperado pero no destaca.",
    # Agrega frases del contexto de tu proyecto:
    "cafe bueno",
    "cafe malo",
]

for frase in frases_a_clasificar:
    r = clasificador(frase)[0]
    print(f"  {r['label']} ({r['score']:.0%})  ->  '{frase[:60]}'")
```

**Observa**: cada frase recibe una etiqueta (1-5 estrellas) y una probabilidad. Fíjate que las respuestas son **iguales cada vez** que ejecutes la celda. Esto es característico de los modelos analíticos: son deterministas.

## Ejercicio 3 — Modelo generativo (produce texto nuevo)

Ahora al revés: le damos un inicio de frase y el modelo la continua.

```python
generador = pipeline("text-generation",
                     model="datificate/gpt2-small-spanish",
                     max_new_tokens=60)
print("Generador listo.")
```

### TU TURNO — prueba con un prompt de tu proyecto

```python
prompt = "hay distintos metodos para hacer cafe"

# Ejecuta esta celda VARIAS veces y ve como cambia la respuesta
resultado = generador(prompt, num_return_sequences=1,
                     do_sample=True, temperature=0.8)
print(resultado[0]['generated_text'])
```
